```python
# =============================================================================
# THE "GRAND MASTER" MINERVINI/O'NEIL SEPA SCANNER
# Coverage: S&P 500 + Nasdaq-100 + Russell 2000 (~2,000 stocks)
#
# HOW TO RUN:
#   1. Go to colab.research.google.com
#   2. File -> New Notebook
#   3. Paste this entire script into one code cell
#   4. Adjust USER SETTINGS below
#   5. Click Run (triangle button) — takes 20-30 minutes
# =============================================================================

!pip install yfinance pandas lxml requests > /dev/null 2>&1

print("Libraries installed. Initializing Grand Master Scanner...")

import yfinance as yf
import pandas as pd
import requests
import warnings
import numpy as np
import logging
from datetime import datetime, timedelta
import time

# Suppress yfinance error messages
logging.getLogger('yfinance').setLevel(logging.CRITICAL)

# --- USER SETTINGS ---
ACCOUNT_SIZE     = 10000    # Your total account size
RISK_PER_TRADE   = 0.0125   # Risk 1.25% of account per trade
MAX_STOP_ALLOWED = 0.10     # Flag stops wider than 10%
ELITE_MODE       = True     # True  = strict (RS>85, VDU+Fundamentals required, VCP informational)
                            # False = standard scan (RS>70, shows all candidates)
MIN_PRICE        = 5.00     # Filter out penny stocks below this price
MIN_AVG_VOLUME   = 100000   # Minimum 50-day average daily share volume (liquidity filter)

# --- KNOWN BAD TICKERS (delisted or unavailable) ---
EXCLUDED_TICKERS = {
    'CAD', 'CRDA', 'INH', 'XTSLA', 'PDLI', 'MOGA', 'MSFUT',
    'SBT', 'THRD', 'GTXI', 'AKE', 'GEFB', 'ADRO', 'BXC', 'DASH'
}

warnings.simplefilter(action='ignore', category=FutureWarning)
warnings.simplefilter(action='ignore', category=pd.errors.PerformanceWarning)

# =============================================================================
# UTILITY: SAFE .info FETCH WITH RETRY
# =============================================================================

def safe_get_info(ticker_obj, retries=3, delay=2):
    """Fetch .info with retries to handle transient Yahoo Finance failures."""
    for attempt in range(retries):
        try:
            info = ticker_obj.info
            # Validate we got real data, not an empty shell
            if info and len(info) > 5:
                return info
        except Exception:
            pass
        time.sleep(delay)
    return {}

# =============================================================================
# UTILITY: STRIP INCOMPLETE INTRADAY BAR
# =============================================================================

def clean_history(df):
    """
    Drop the most recent bar if it was fetched during market hours
    (i.e., today's partial candle). This prevents incomplete volume/price
    data from corrupting RS scores, VDU, and trend calculations.
    """
    if df is None or len(df) == 0:
        return df
    now_utc  = pd.Timestamp.utcnow()
    last_bar = df.index[-1]
    # Convert last bar to UTC for comparison
    if last_bar.tzinfo is None:
        last_bar_utc = last_bar.tz_localize('UTC')
    else:
        last_bar_utc = last_bar.tz_convert('UTC')
    # If the last bar is today, drop it — it may be incomplete
    if last_bar_utc.date() == now_utc.date():
        df = df.iloc[:-1]
    return df

# =============================================================================
# PHASE 0: MARKET CONTEXT (THE "M" FACTOR)
# =============================================================================

print("\n[Phase 0] DIAGNOSING MARKET CONTEXT...")

market_signal = "GREEN (Aggressive)"
sp_trend = nq_trend = rut_trend = "N/A"
sp_dist = nq_dist = rut_dist = 0

try:
    indices = yf.download(["^GSPC", "^IXIC", "^RUT"], period="80d", progress=False)

    def check_index(ticker):
        try:
            try:
                closes = indices['Close'][ticker]
                vols   = indices['Volume'][ticker]
            except:
                closes = indices['Close']
                vols   = indices['Volume']

            avg_vol_50 = vols.rolling(50).mean()

            dist_days = 0
            for i in range(len(closes) - 25, len(closes)):
                if i < 1: continue
                pct = (closes.iloc[i] - closes.iloc[i-1]) / closes.iloc[i-1]
                if pct < -0.002 and vols.iloc[i] > avg_vol_50.iloc[i]:
                    dist_days += 1

            for i in range(len(closes) - 25, len(closes)):
                if i < 1: continue
                pct = (closes.iloc[i] - closes.iloc[i-1]) / closes.iloc[i-1]
                if pct >= 0.017 and vols.iloc[i] > avg_vol_50.iloc[i]:
                    dist_days = max(0, dist_days - 1)

            trend = "UP" if closes.iloc[-1] > closes.tail(50).mean() else "DOWN"
            return trend, dist_days
        except:
            return "N/A", 0

    sp_trend,  sp_dist  = check_index("^GSPC")
    nq_trend,  nq_dist  = check_index("^IXIC")
    rut_trend, rut_dist = check_index("^RUT")

    max_dist = max(sp_dist, nq_dist)

    if max_dist >= 6:
        market_signal = "RED (Heavy Distribution)"
    elif max_dist >= 4:
        market_signal = "YELLOW (Caution)"

    if sp_trend == "DOWN" or nq_trend == "DOWN":
        if market_signal == "GREEN (Aggressive)":
            market_signal = "YELLOW (Trend Broken)"

    if sp_trend == "UP" and rut_trend == "DOWN":
        if "GREEN" in market_signal:
            market_signal = "YELLOW (Divergence - Small Caps Lagging)"

    print(f"   > S&P 500:      {sp_trend} | Dist Days (25-session): {sp_dist}")
    print(f"   > Nasdaq:       {nq_trend} | Dist Days (25-session): {nq_dist}")
    print(f"   > Russell 2000: {rut_trend} | Dist Days (25-session): {rut_dist}")
    print(f"\n   >>> MARKET SIGNAL: {market_signal} <<<")

except Exception as e:
    print(f"   [WARNING] Market context check failed: {e}")

if market_signal.startswith("RED"):
    print("\n" + "!"*80)
    print("  MARKET SIGNAL IS RED - HIGH DISTRIBUTION DETECTED.")
    print("  Minervini/O'Neil methodology: Move to CASH. Do NOT initiate new positions.")
    print("  Scan will run for research purposes only. DO NOT act on these signals.")
    print("!"*80)
elif "YELLOW" in market_signal:
    print("\n  [CAUTION] Market is YELLOW. Reduce position sizes.")
    print("  Use pilot positions only. Raise cash on weakness in existing holdings.\n")

# =============================================================================
# PHASE 1: TICKER COLLECTION (S&P 500 + Nasdaq-100 + Russell 2000)
# =============================================================================

def get_tickers():
    tickers = set()
    headers = {'User-Agent': 'Mozilla/5.0'}

    # --- Source 1: S&P 500 (Wikipedia) ---
    try:
        url   = 'https://en.wikipedia.org/wiki/List_of_S%26P_500_companies'
        sp500 = pd.read_html(requests.get(url, headers=headers).text)[0]['Symbol'].tolist()
        tickers.update(sp500)
        print(f"   > S&P 500:    {len(sp500)} tickers loaded")
    except Exception as e:
        print(f"   > S&P 500 FAILED: {e}")

    # --- Source 2: Nasdaq-100 (Wikipedia) — robust table finder ---
    # FIX: Do NOT hardcode table index [4]. Wikipedia reorganizes tables
    # periodically. Scan all tables for one that has a 'Ticker' column.
    try:
        url    = 'https://en.wikipedia.org/wiki/Nasdaq-100'
        tables = pd.read_html(requests.get(url, headers=headers).text)
        nq100  = []
        for tbl in tables:
            # Normalize column names for comparison
            cols = [str(c).strip() for c in tbl.columns]
            if 'Ticker' in cols:
                nq100 = tbl['Ticker'].dropna().tolist()
                if len(nq100) > 50:   # Must look like a real constituent list
                    break
        if nq100:
            tickers.update(nq100)
            print(f"   > Nasdaq-100: {len(nq100)} tickers loaded")
        else:
            raise ValueError("No table with 'Ticker' column and 50+ rows found")
    except Exception as e:
        print(f"   > Nasdaq-100 FAILED: {e}")

    # --- Source 3: Russell 2000 via iShares IWM holdings ---
    try:
        url  = "https://www.ishares.com/us/products/239710/ISHARES-RUSSELL-2000-ETF/1467271812596.ajax?fileType=csv&fileName=IWM_holdings&dataType=fund"
        resp = requests.get(url, headers=headers, timeout=30)
        lines = resp.text.split('\n')
        header_idx = None
        for idx, line in enumerate(lines):
            if len(line) < 200 and ('Ticker' in line or 'ticker' in line):
                header_idx = idx
                break
        if header_idx is None:
            raise ValueError("Could not locate Ticker header row in IWM CSV")
        from io import StringIO
        df_iwm = pd.read_csv(StringIO('\n'.join(lines[header_idx:])), on_bad_lines='skip')
        ticker_col = None
        for col in df_iwm.columns:
            if col.strip().lower() == 'ticker':
                ticker_col = col
                break
        if ticker_col is None:
            raise ValueError("Ticker column not found in IWM CSV")
        iwm_raw     = df_iwm[ticker_col].dropna().tolist()
        iwm_tickers = []
        for t in iwm_raw:
            t_clean = str(t).strip().lstrip('$')
            if t_clean and t_clean not in ('-', 'nan', 'CASH', 'USD', ''):
                if len(t_clean) <= 6 and t_clean.replace('-','').isalpha():
                    iwm_tickers.append(t_clean)
        tickers.update(iwm_tickers)
        print(f"   > Russell 2000 (IWM): {len(iwm_tickers)} tickers loaded")
    except Exception as e:
        print(f"   > Russell 2000 (IWM) FAILED: {e}")
        try:
            url2   = "https://www.ishares.com/us/products/239710/ISHARES-RUSSELL-2000-ETF/1467271812596.ajax?fileType=csv&fileName=IWM_holdings&dataType=fund&asOfDate="
            resp2  = requests.get(url2, headers=headers, timeout=30)
            lines2 = resp2.text.split('\n')
            header_idx2 = next((i for i, l in enumerate(lines2)
                                if len(l) < 200 and 'Ticker' in l), None)
            if header_idx2:
                df2  = pd.read_csv(StringIO('\n'.join(lines2[header_idx2:])), on_bad_lines='skip')
                col2 = next((c for c in df2.columns if c.strip().lower() == 'ticker'), None)
                if col2:
                    fb = [str(t).strip().lstrip('$') for t in df2[col2].dropna()
                          if len(str(t).strip().lstrip('$')) <= 6
                          and str(t).strip().lstrip('$').replace('-','').isalpha()]
                    tickers.update(fb)
                    print(f"   > Russell 2000 (fallback): {len(fb)} tickers loaded")
        except Exception as e2:
            print(f"   > Russell 2000: All sources failed ({e2}). Continuing with S&P+Nasdaq only.")

    # Clean tickers and apply exclusion list
    cleaned = []
    seen    = set()
    for t in tickers:
        if not isinstance(t, str):
            continue
        t = t.strip().lstrip('$').upper()
        t = t.replace('.', '-')
        if t in ('', '-', 'NAN', 'CASH', 'USD') or t in seen:
            continue
        if t in EXCLUDED_TICKERS:
            continue
        if len(t) < 1 or len(t) > 6:
            continue
        parts = t.split('-')
        if not all(p.isalpha() for p in parts if p):
            continue
        seen.add(t)
        cleaned.append(t)

    print(f"\n   > TOTAL UNIQUE TICKERS: {len(cleaned)}")
    return cleaned

print("\n[Phase 1] COLLECTING TICKERS...")
tickers = get_tickers()

print(f"\n[Phase 1] DOWNLOADING {len(tickers)} STOCKS (estimated 20-30 minutes)...")
print("   Note: Progress updates every 50 stocks.\n")

# FIX: Apply clean_history() to SPY as well so its bar count stays in sync
spy_hist = yf.Ticker("SPY").history(
    start=(datetime.now() - timedelta(days=540)).strftime('%Y-%m-%d'), interval="1d"
)
spy_hist = clean_history(spy_hist)

start_date  = (datetime.now() - timedelta(days=540)).strftime('%Y-%m-%d')
stock_cache = {}
rs_list     = []
skipped_liquidity = 0
skipped_history   = 0

for i, ticker in enumerate(tickers):
    if i % 50 == 0:
        print(f"   > {i}/{len(tickers)} processed | {len(stock_cache)} passed filters...", end="\r")

    try:
        df = yf.Ticker(ticker).history(start=start_date, interval="1d")

        if isinstance(df.columns, pd.MultiIndex):
            df.columns = df.columns.get_level_values(0)

        # FIX: Strip today's potentially incomplete bar before any calculations
        df = clean_history(df)

        if len(df) < 252:
            skipped_history += 1
            continue

        current_price = df['Close'].iloc[-1]
        avg_volume    = df['Volume'].rolling(50).mean().iloc[-1]

        if current_price < MIN_PRICE:
            skipped_liquidity += 1
            continue
        if avg_volume < MIN_AVG_VOLUME:
            skipped_liquidity += 1
            continue

        stock_cache[ticker] = df

        c     = df['Close'].iloc[-1]
        c_63  = df['Close'].iloc[-63]
        c_126 = df['Close'].iloc[-126]
        c_189 = df['Close'].iloc[-189]
        c_252 = df['Close'].iloc[-252]

        if c_63 == 0: continue

        score = (0.4 * ((c - c_63)  / c_63)  +
                 0.2 * ((c - c_126) / c_126)  +
                 0.2 * ((c - c_189) / c_189)  +
                 0.2 * ((c - c_252) / c_252))

        rs_list.append({'Ticker': ticker, 'Raw_Score': score})

        if i % 100 == 0 and i > 0:
            time.sleep(0.5)

    except:
        continue

print(f"\n\n   > Download complete.")
print(f"   > Stocks passing price/volume filters: {len(stock_cache)}")
print(f"   > Skipped (insufficient history):      {skipped_history}")
print(f"   > Skipped (price/volume filters):      {skipped_liquidity}")

rs_df             = pd.DataFrame(rs_list).dropna(subset=['Raw_Score'])
rs_df['Rank_Pct'] = rs_df['Raw_Score'].rank(pct=True)
rs_df['RS_Rating']= (rs_df['Rank_Pct'] * 99).fillna(0).astype(int)
rs_map            = rs_df.set_index('Ticker')['RS_Rating'].to_dict()

print(f"   > RS ratings calculated for {len(rs_map)} stocks.")

# =============================================================================
# PHASE 1.5: MARKET BREADTH DASHBOARD
# =============================================================================

print("\n\n[Phase 1.5] CALCULATING MARKET BREADTH...")

total_stocks = len(stock_cache)
above_50 = above_200 = up_vol = down_vol = new_highs = new_lows = 0

for ticker, df in stock_cache.items():
    try:
        curr    = df['Close'].iloc[-1]
        prev    = df['Close'].iloc[-2]
        vol     = df['Volume'].iloc[-1]
        high_52 = df['Close'].tail(252).max()
        low_52  = df['Close'].tail(252).min()
        sma_50  = df['Close'].rolling(50).mean().iloc[-1]
        sma_200 = df['Close'].rolling(200).mean().iloc[-1]

        if curr > sma_50:  above_50  += 1
        if curr > sma_200: above_200 += 1
        if curr >= (high_52 * 0.99):  new_highs += 1
        elif curr <= (low_52 * 1.01): new_lows  += 1
        if curr > prev:   up_vol   += vol
        elif curr < prev: down_vol += vol
    except:
        continue

pct_above_50  = (above_50  / total_stocks) * 100 if total_stocks else 0
pct_above_200 = (above_200 / total_stocks) * 100 if total_stocks else 0
up_down_ratio = (up_vol / down_vol) if down_vol > 0 else 99.0
trend_broken  = (sp_trend == "DOWN" or nq_trend == "DOWN")

print(f"   > Universe size:       {total_stocks} stocks")
print(f"   > Stocks > 50-Day MA:  {pct_above_50:.1f}%")
print(f"   > Stocks > 200-Day MA: {pct_above_200:.1f}%")
print(f"   > Up/Down Vol Ratio:   {up_down_ratio:.2f}  (Ideal > 1.0)")
print(f"   > New Highs / Lows:    {new_highs} Highs  vs  {new_lows} Lows")

if new_lows > new_highs:
    breadth_status = "RED (New Lows Dominating)"
elif up_down_ratio < 1.0 and pct_above_50 < 40:
    breadth_status = "RED (Broad Weakness)"
elif up_down_ratio > 1.0 and pct_above_50 > 50:
    if trend_broken:
        breadth_status = "YELLOW (Speculative Bounce - Primary Trend Down)"
    else:
        breadth_status = "GREEN (Accumulation - Broad Participation)"
else:
    breadth_status = "YELLOW (Mixed Internals)"

print(f"   > INTERNAL SIGNAL: {breadth_status}")

# =============================================================================
# PHASE 1.8: PRE-FETCH FUNDAMENTALS + SECTOR/INDUSTRY + MARKET CAP
# =============================================================================

print(f"\n[Phase 1.8] PRE-FETCHING FUNDAMENTALS + SECTOR/INDUSTRY + MARKET CAP for {len(stock_cache)} stocks...")
print("   (This is the slow step — be patient...)")

fund_map   = {}
sector_map = {}
mktcap_map = {}

def format_mktcap(val):
    try:
        val = float(val)
        if val >= 1_000_000_000_000:
            return f"${val/1_000_000_000_000:.1f}T"
        elif val >= 1_000_000_000:
            return f"${val/1_000_000_000:.1f}B"
        elif val >= 1_000_000:
            return f"${val/1_000_000:.0f}M"
        else:
            return f"${val:,.0f}"
    except:
        return "N/A"

for i, ticker in enumerate(stock_cache.keys()):
    if i % 50 == 0:
        print(f"   > {i}/{len(stock_cache)} fetched...", end="\r")

    try:
        t_obj = yf.Ticker(ticker)

        # --- Fundamentals ---
        fin       = t_obj.quarterly_financials
        eps_flag  = False
        eps_accel = False
        rev_flag  = False

        if "Net Income" in fin.index:
            ni = fin.loc["Net Income"]
            if len(ni) > 4 and ni.iloc[4] != 0:
                growth_q0 = (ni.iloc[0] - ni.iloc[4]) / abs(ni.iloc[4])
                if growth_q0 > 0.25:
                    eps_flag = True
                    if len(ni) >= 6 and ni.iloc[5] != 0:
                        growth_q1 = (ni.iloc[1] - ni.iloc[5]) / abs(ni.iloc[5])
                        if growth_q0 > growth_q1:
                            eps_accel = True

        if "Total Revenue" in fin.index:
            rev = fin.loc["Total Revenue"]
            if len(rev) > 4 and rev.iloc[4] != 0:
                if ((rev.iloc[0] - rev.iloc[4]) / abs(rev.iloc[4])) > 0.20:
                    rev_flag = True

        notes = []
        if eps_flag:
            notes.append("EPS++" if eps_accel else "EPS+")
        if rev_flag:
            notes.append("Sale+")
        fund_map[ticker] = " ".join(notes) if notes else "Lag"

        # --- Sector / Industry / Market Cap via safe retry ---
        info = safe_get_info(t_obj)
        if info:
            sector   = info.get('sector', 'N/A')
            industry = info.get('industry', 'N/A')
            sector_abbrev = {
                'Technology': 'Tech',
                'Healthcare': 'Health',
                'Financial Services': 'Finance',
                'Consumer Cyclical': 'Cons.Cyc',
                'Consumer Defensive': 'Cons.Def',
                'Communication Services': 'Comm.Svc',
                'Basic Materials': 'Materials',
                'Real Estate': 'RE',
                'Industrials': 'Indust.',
                'Energy': 'Energy',
                'Utilities': 'Utilities',
            }
            sector_short       = sector_abbrev.get(sector, sector)
            sector_map[ticker] = f"{sector_short} | {industry}"
            mktcap_map[ticker] = format_mktcap(info.get('marketCap'))
        else:
            sector_map[ticker] = "N/A"
            mktcap_map[ticker] = "N/A"

        if i % 50 == 0 and i > 0:
            time.sleep(1)

    except:
        fund_map[ticker]   = "N/A"
        sector_map[ticker] = "N/A"
        mktcap_map[ticker] = "N/A"

print(f"\n   > Fundamentals + Sector + Market Cap cached for {len(fund_map)} stocks.")

# =============================================================================
# RSI HELPER FUNCTION (14-period, Wilder's smoothing)
# =============================================================================

def calc_rsi(closes, period=14):
    try:
        delta    = closes.diff()
        gain     = delta.clip(lower=0)
        loss     = -delta.clip(upper=0)
        avg_gain = gain.ewm(alpha=1/period, min_periods=period, adjust=False).mean()
        avg_loss = loss.ewm(alpha=1/period, min_periods=period, adjust=False).mean()
        rs  = avg_gain / avg_loss.replace(0, np.nan)
        rsi = 100 - (100 / (1 + rs))
        return round(float(rsi.iloc[-1]), 1)
    except:
        return None

# =============================================================================
# PHASE 2: THE SCANNER
# =============================================================================

rs_threshold = 85 if ELITE_MODE else 70
mode_label   = "ELITE SNIPER" if ELITE_MODE else "GRAND MASTER"

print(f"\n[Phase 2] RUNNING {mode_label} SCREEN  (RS threshold: {rs_threshold})")

if market_signal.startswith("RED"):
    print("  *** RED MARKET - RESULTS FOR RESEARCH ONLY. DO NOT BUY. ***")

print("-" * 160)
print(f"{'TICKER':<7} | {'RS':<3} | {'RSI':<5} | {'RS_NH':<5} | {'VCP':<3} | {'VDU':<7} | {'SQUAT':<5} | "
      f"{'FUNDS':<10} | {'PRICE':<8} | {'BUY >':<8} | {'STOP <':<8} | {'RISK %':<7} | {'SHARES':<6} | "
      f"{'INDEX':<8} | {'MKT CAP':<9} | {'SECTOR | INDUSTRY'}")
print("-" * 160)

# Index membership lookup
sp500_set = set()
nq100_set = set()
try:
    url = 'https://en.wikipedia.org/wiki/List_of_S%26P_500_companies'
    sp500_set = set(pd.read_html(requests.get(url, headers={'User-Agent': 'Mozilla/5.0'}).text)[0]['Symbol'].tolist())
except: pass
try:
    url    = 'https://en.wikipedia.org/wiki/Nasdaq-100'
    tables = pd.read_html(requests.get(url, headers={'User-Agent': 'Mozilla/5.0'}).text)
    for tbl in tables:
        cols = [str(c).strip() for c in tbl.columns]
        if 'Ticker' in cols:
            candidates = tbl['Ticker'].dropna().tolist()
            if len(candidates) > 50:
                nq100_set = set(candidates)
                break
except: pass

passing_stocks = []
d_rs = d_trend = d_vdu = d_funds = d_other = 0

for ticker, data in stock_cache.items():
    try:
        rs = rs_map.get(ticker, 0)
        if rs < rs_threshold:
            d_rs += 1
            continue

        close   = data['Close'].iloc[-1]
        sma_50  = data['Close'].rolling(50).mean().iloc[-1]
        sma_150 = data['Close'].rolling(150).mean().iloc[-1]
        sma_200 = data['Close'].rolling(200).mean().iloc[-1]
        low_52  = data['Close'].tail(252).min()
        high_52 = data['Close'].tail(252).max()

        # Trend Template
        if not (close > sma_50 and close > sma_150 and close > sma_200 and
                sma_50 > sma_150 and sma_150 > sma_200):
            d_trend += 1
            continue
        if not (close > (1.3 * low_52) and close > (0.75 * high_52)):
            d_trend += 1
            continue

        sma_200_20_ago = data['Close'].rolling(200).mean().shift(20).iloc[-1]
        if sma_200 <= sma_200_20_ago:
            d_trend += 1
            continue

        # VCP Detection — informational only
        std_5  = data['Close'].rolling(5).std().iloc[-1]
        std_20 = data['Close'].rolling(20).std().iloc[-1]
        std_60 = data['Close'].rolling(60).std().iloc[-1]

        if std_20 == 0 or std_60 == 0:
            d_other += 1
            continue

        tight_now   = (std_5 / std_20) < 0.55
        contracting = (std_20 / std_60) < 0.85
        vcp = "YES" if (tight_now and contracting) else "No"

        # Squat Detection
        squat_display = "-"
        try:
            for i in range(1, 3):
                day_high  = data['High'].iloc[-i]
                day_low   = data['Low'].iloc[-i]
                day_close = data['Close'].iloc[-i]
                recent_hi = data['High'].iloc[-i-10:-i].max()
                if day_high > recent_hi:
                    rng = day_high - day_low
                    if rng > 0 and ((day_close - day_low) / rng) < 0.50:
                        squat_display = "YES"
                        break
        except: pass

        # Volume Dry-Up
        curr_vol = data['Volume'].iloc[-1]
        avg_vol  = data['Volume'].rolling(50).mean().iloc[-1]
        if curr_vol < (avg_vol * 0.5):  vdu_display = "Extreme"
        elif curr_vol < avg_vol:         vdu_display = "Yes"
        else:                            vdu_display = "No"

        if ELITE_MODE and vdu_display == "No":
            d_vdu += 1
            continue

        # RS Line New High — 50-day window
        try:
            aligned_spy  = spy_hist['Close'].reindex(data.index).ffill()
            rs_line      = data['Close'] / aligned_spy
            rs_line_high = rs_line.tail(50).max()
            rs_new_high  = "YES" if rs_line.iloc[-1] >= (rs_line_high * 0.99) else "-"
        except:
            rs_new_high = "-"

        # RSI (14-period)
        rsi_val     = calc_rsi(data['Close'])
        rsi_display = f"{rsi_val:.1f}" if rsi_val is not None else "N/A"

        # Fundamentals
        fund_display = fund_map.get(ticker, "N/A")
        if ELITE_MODE and "EPS+" not in fund_display and "Sale+" not in fund_display:
            d_funds += 1
            continue

        # Execution
        recent_high    = data['High'].tail(10).max()
        buy_trigger    = recent_high * 1.005
        natural_stop   = data['Low'].tail(10).min()
        risk_per_share = buy_trigger - natural_stop

        if risk_per_share <= 0:
            d_other += 1
            continue

        dollar_risk = ACCOUNT_SIZE * RISK_PER_TRADE
        shares      = int(dollar_risk / risk_per_share)
        risk_pct    = risk_per_share / buy_trigger
        risk_str    = f"{risk_pct*100:.1f}%"
        if risk_pct > MAX_STOP_ALLOWED: risk_str = "!" + risk_str

        # Index label
        t_clean = ticker.replace('-', '.')
        if t_clean in sp500_set or ticker in sp500_set:
            index_label = "S&P500"
        elif t_clean in nq100_set or ticker in nq100_set:
            index_label = "NQ100"
        else:
            index_label = "RUT2000"

        mktcap_display = mktcap_map.get(ticker, "N/A")
        sector_display = sector_map.get(ticker, "N/A")

        print(f"{ticker:<7} | {rs:<3} | {rsi_display:<5} | {rs_new_high:<5} | {vcp:<3} | {vdu_display:<7} | "
              f"{squat_display:<5} | {fund_display:<10} | ${close:<7.2f} | ${buy_trigger:<7.2f} | "
              f"${natural_stop:<7.2f} | {risk_str:<7} | {shares:<6} | {index_label:<8} | "
              f"{mktcap_display:<9} | {sector_display}")

        passing_stocks.append(ticker)

    except:
        d_other += 1
        continue

print("-" * 160)
print(f"\n{mode_label} SCAN COMPLETE.  Found {len(passing_stocks)} candidates from ~{len(stock_cache)} stocks screened.")

print(f"\n  [DIAGNOSTIC] Filter elimination breakdown:")
print(f"   > Failed RS threshold ({rs_threshold}+):   {d_rs}")
print(f"   > Failed Trend Template:          {d_trend}")
print(f"   > Failed VDU (volume too high):   {d_vdu}")
print(f"   > Failed Fundamentals:            {d_funds}")
print(f"   > Other (data/calculation):       {d_other}")
print(f"   > Passed all filters:             {len(passing_stocks)}")

if market_signal.startswith("RED"):
    print("\n  *** REMINDER: MARKET IS RED. DO NOT initiate new positions. ***")
elif "YELLOW" in market_signal:
    print("\n  *** REMINDER: MARKET IS YELLOW. Use reduced/pilot position sizes only. ***")

# =============================================================================
# LEGEND
# =============================================================================

print("\n" + "="*80)
print(f"  {mode_label} DASHBOARD LEGEND")
print("="*80)
print("UNIVERSE")
print(f"  Coverage: S&P 500 + Nasdaq-100 + Russell 2000")
print(f"  Filters:  Price > ${MIN_PRICE:.2f}  |  Avg Volume > {MIN_AVG_VOLUME:,}")
print("-"*80)
print("MARKET SIGNALS")
print("  GREEN  : Conditions favorable. Normal position sizing.")
print("  YELLOW : Caution. Use pilot (half) positions. Raise stops.")
print("  RED    : Heavy distribution. Move to cash. No new buys.")
print("  NOTE   : Strong rally days (up 1.7%+ on high vol) erase one dist. day (O'Neil rule)")
print("-"*80)
print("FUNDAMENTALS")
print("  EPS++  : Net Income >25% YoY most recent quarter AND accelerating")
print("  EPS+   : Net Income >25% YoY most recent quarter")
print("  Sale+  : Revenue growth > 20% YoY (most recent quarter)")
print("  Lag    : Technicals pass but fundamentals weak - proceed with caution")
print("-"*80)
print("TECHNICALS")
print("  RSI    : 14-period RSI (Wilder smoothing). 50-70 = healthy trend; >70 = extended")
print("  RS_NH  : YES = RS Line at new 50-day high | GREEN market = confirmed leadership signal")
print("           | RED/YELLOW market = use for watchlist research, verify on chart")
print("  VCP    : YES = Math suggests volatility contraction | verify on TradingView with VCS")
print("           NO  = Math does not detect contraction | still check chart — proxy is imperfect")
print("  VDU    : Extreme = <50% avg vol | Yes = below avg | No = above avg")
print("  SQUAT  : YES = Bearish reversal candle - AVOID this setup")
print("  INDEX  : Which index the stock belongs to")
print("  MKT CAP: T = Trillion | B = Billion | M = Million")
print("  SECTOR : Sector and industry classification from Yahoo Finance")
print("-"*80)
print("EXECUTION")
print("  BUY >  : Place a Buy Stop Limit order above this price")
print("  STOP < : Hard stop loss - do not lower this under any circumstance")
print("  RISK % : Distance from buy trigger to stop (! = exceeds 10% max)")
print(f"  SHARES : Position size for {RISK_PER_TRADE*100:.1f}% account risk (${ACCOUNT_SIZE*RISK_PER_TRADE:.0f} at risk)")
print("="*80)
```
